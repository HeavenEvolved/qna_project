```python
from langchain.document_loaders import UnstructuredPDFLoader
from unstructured.cleaners.core import clean_non_ascii_chars, clean, remove_punctuation
```

```python
docs = UnstructuredPDFLoader('./files/34829000006448425.pdf', mode="paged", post_processors=[clean_non_ascii_chars, remove_punctuation, lambda x: clean(x, bullets=True, extra_whitespace=True, dashes=True, trailing_punctuation=True, lowercase=True)]).load()
```

```python
print(docs[0].metadata['source'])
```

```python
for page in docs:
    print(page.page_content)
```
